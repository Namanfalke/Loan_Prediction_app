```python
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler 
from sklearn.preprocessing import LabelEncoder  
import joblib
```

```python
df = pd.read_csv('../clean data/cleaned_data.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.duplicated().sum()
```

```python
df['Age']=df['Age'].astype(int)
```

```python
df.head()
```

```python
X = df[['Annual_Income', 'Credit_Score', 'Loan_Amount', 'Dependents', 'Education']]
```

```python
y = df['Loan_Status']
```

```python
le=LabelEncoder()
```

```python
Labeled_Education = le.fit_transform(X['Education'])
Labeled_Education=pd.Series(Labeled_Education)
Labeled_Education.unique()
```

```python
X['Labelled_Education'] = Labeled_Education
X.drop('Education', axis=1, inplace=True)
```

```python
X.head()
#not_graduated == 1
#graduated == 0
```

```python
ss=StandardScaler()
Scalled_X=ss.fit_transform(X)
```

```python
# Scalled_X = pd.DataFrame(Scalled_X, columns=X.columns)
Scalled_X = pd.DataFrame(Scalled_X, columns=[['Annual_Income','Credit_Score','Loan_Amount','Dependents','Education']])
```

```python
Scalled_X.head()
```

```python
# joblib.dump(ss,'StandardScaler.pkl')
```

```python
Scalled_X.head()
```

```python
y.head()
```

```python
X_train,X_test,y_train,y_test=train_test_split(Scalled_X,y,test_size=0.2,random_state=42)
```

```python
knn=KNeighborsClassifier(n_neighbors=5)
knn
```

```python
knn.fit(X_train,y_train)
```

```python
knn.score(X_test,y_test)
```

```python
knn.predict(X_test)
```
