```python
import numpy as np
import pandas as pd
import openpyxl
import joblib
```

```python
df=pd.read_excel("../raw data/dd.xlsx")
df.head()
```

```python
df.drop(columns=['Customer_ID'], inplace=True)
```

```python
df.info()
```

```python
df.columns
```

```python
df.dtypes
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.isna().sum()
```

```python
df['Age']=df['Age'].fillna(df['Age'].mean())
df.isna().sum()
```

```python
df['Annual_Income']=df['Annual_Income'].fillna(df['Annual_Income'].mean())
df.isna().sum()
```

```python
df['Credit_Score']=df['Credit_Score'].fillna(df['Credit_Score'].mean())
df.isna().sum()
```

```python
df['Loan_Amount']=df['Loan_Amount'].fillna(df['Loan_Amount'].mean())
df.isna().sum()
```

```python
df['Education']=df['Education'].ffill()
df.isna().sum()
```

```python
df['Self_Employed']=df['Self_Employed'].bfill()
df.isna().sum()
```

```python
df.info()
```

```python
df.head()
```

```python
df['Education']=df['Education'].str.lower()
df['Education']=df['Education'].str.strip()
```

```python
df.head()
```

```python
df.to_csv("../clean data/cleaned_data.csv", index=False)
```
